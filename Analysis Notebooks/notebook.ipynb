```python
df = pd.read_csv(
    "occurrences.csv",
    sep=";",
    dtype=str,
    low_memory=False,
    encoding="utf-8"
)

df.head()
```

```python
df.shape
df.head(3)
df.info()
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(r"[^a-z0-9]+", "_", regex=True)
      .str.strip("_")
)
df.columns.tolist()
```

```python
date_cols = [
    "local_date_main_event",
    "publication_date",
    "occurrence_record_modified_on",
    "last_updated_at",
    "occurrences_reported_from",
    "occurrences_reported_to"
]

for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")

for col in ["latitude", "longitude", "year_reported"]:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.isna().sum().sort_values(ascending=False).head(15)
df["occurrence_severity"].value_counts(dropna=False).head(10)
df["main_event_l1"].value_counts(dropna=False).head(10)
```

```python
df.to_csv("occurrences_cleaned.csv", index=False)
```
